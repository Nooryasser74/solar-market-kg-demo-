## 1. Setup and load every inspection file


````python
from pathlib import Path
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
INSPECTION = ROOT / "inspection"

print("Notebook folder:", ROOT)
print("Inspection folder:", INSPECTION)
print("Inspection folder exists:", INSPECTION.exists())

if not INSPECTION.exists():
    raise FileNotFoundError(
        "The inspection folder was not found. "
        "Place this notebook inside pilot_01."
    )
````

````python
# Load all CSV files automatically
datasets = {}

for path in sorted(INSPECTION.glob("*.csv")):
    try:
        datasets[path.name] = pd.read_csv(path, encoding="utf-8-sig")
    except UnicodeDecodeError:
        datasets[path.name] = pd.read_csv(path)

print(f"Loaded {len(datasets)} CSV datasets.\n")

for name, df in datasets.items():
    print(f"{name:40s} {len(df):>6} rows x {len(df.columns):>3} columns")
````

## 2. Read the summary file


````python
summary_path = INSPECTION / "00_summary.txt"

if summary_path.exists():
    print(summary_path.read_text(encoding="utf-8"))
else:
    print("00_summary.txt was not found.")
````

## 3. Helper: view any dataset by filename


````python
def view(filename, rows=20):
    if filename not in datasets:
        print("Dataset not found:", filename)
        print("\nAvailable datasets:")
        for name in datasets:
            print(" -", name)
        return

    df = datasets[filename]

    print(filename)
    print(f"Rows: {len(df)}")
    print(f"Columns: {len(df.columns)}")
    print("\nColumn names:")
    print(list(df.columns))
    print()

    display(df.head(rows))
````

## 17. Search entities by word

Change `search_term` to anything you want, for example:
- `photovoltaik`
- `eag`
- `österreich`
- `netz`
- `förder`


````python
search_term = "photovoltaik"

df = datasets["03_entity_review.csv"]

mask = df.astype(str).apply(
    lambda col: col.str.contains(search_term, case=False, na=False)
).any(axis=1)

display(df[mask])
````
