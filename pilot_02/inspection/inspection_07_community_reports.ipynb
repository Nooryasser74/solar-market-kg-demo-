# Pilot 2 Community-Report Audit

## Objective

This inspection evaluated whether the community reports generated by
GraphRAG faithfully summarized the underlying Austrian Photovoltaic
Strategy and its extracted knowledge graph.

Community reports must not be treated as source evidence. They are
LLM-generated interpretations constructed from extracted entities,
relationships, and community structures.

## Audit design

Pilot 2 produced 110 community reports, compared with 62 reports in
Pilot 1.

This represents:

- an absolute increase of 48 reports;
- a relative increase of 77.4%.

A systematic sample of 13 Pilot 2 community reports was inspected. The
sample combined:

- the five highest-ranked reports;
- the five reports representing the largest communities;
- seven reproducibly selected random reports;
- removal of duplicate selections.

Each sampled report was assessed for:

- source grounding;
- accuracy of synthesis;
- inherited graph errors;
- claims newly introduced by the report-generating LLM;
- correctness of numerical claims;
- preservation of modality and implementation status;
- unsupported causal language;
- unsupported elaboration.

## Results

### Source grounding

| Grounding classification | Reports | Percentage |
|---|---:|---:|
| HIGH | 2 | 15.4% |
| PARTIAL | 11 | 84.6% |
| LOW | 0 | 0.0% |

All sampled reports contained information connected to the source or
the extracted graph. However, only 15.4% were assessed as highly
grounded throughout. Most reports combined valid source information
with broader interpretations derived from the graph.

### Primary reliability classification

| Classification | Reports | Percentage |
|---|---:|---:|
| ACCURATE_SYNTHESIS | 4 | 30.8% |
| CONTAINS_LLM_INTRODUCED_CLAIMS | 6 | 46.2% |
| CAUSAL_GENERALIZATION | 2 | 15.4% |
| UNSUPPORTED_ELABORATION | 1 | 7.7% |

Only 4 of the 13 reports were primarily classified as accurate
syntheses.

The most common problem was the introduction of claims or
interpretations that were not explicitly present in the source or
could not be safely derived from the graph.

### Inherited graph errors

| Inherited graph error | Reports | Percentage |
|---|---:|---:|
| YES | 11 | 84.6% |
| NO | 2 | 15.4% |

Most community reports inherited at least one problem from the entity
or relationship graph.

These problems included:

- duplicate or unresolved entities;
- broad or imprecise relationship types;
- incorrect or ambiguous relationship endpoints;
- associations presented with greater specificity than the graph
  evidence allowed;
- inconsistent implementation-status information.

This shows that community-report quality is strongly dependent on the
quality of the preceding graph-extraction and entity-resolution
stages.

### New LLM-introduced claims

| New LLM claim | Reports | Percentage |
|---|---:|---:|
| YES | 9 | 69.2% |
| NO | 4 | 30.8% |

Nine reports contained at least one interpretive claim introduced
during report generation.

Therefore, community-report errors cannot be attributed solely to the
underlying graph. The report-generation LLM represents an additional
interpretation layer that may introduce new generalizations.

### Numerical claims

| Numerical assessment | Reports | Percentage |
|---|---:|---:|
| YES | 6 | 46.2% |
| PARTIAL | 1 | 7.7% |
| NOT_APPLICABLE | 6 | 46.2% |

Among the seven reports containing relevant numerical claims, six
presented the principal numerical values correctly and one was only
partially correct.

This indicates that Pilot 2 substantially improved the representation
of targets and market metrics. Important values such as the following
were generally retained:

- at least 11 TWh of additional PV generation by 2030;
- approximately 41 TWh of PV potential by 2040;
- approximately 2.5 GW of additional PV capacity in 2023;
- approximately 6.3 TWh of PV generation in 2023;
- approximately 10% of electricity consumption supplied by PV in
  2023.

Nevertheless, a correct number does not guarantee a correct claim. The
status, responsible policy, comparison baseline, deadline, and
relationship connecting the value to other entities must also be
verified.

### Preservation of modality

| Modality assessment | Reports | Percentage |
|---|---:|---:|
| YES | 5 | 38.5% |
| PARTIAL | 8 | 61.5% |

Only five reports consistently preserved distinctions such as:

- measured outcome;
- explicit fact;
- target;
- scenario;
- recommendation;
- planned action;
- proposal;
- legal requirement.

Eight reports preserved these distinctions only partially. In these
cases, the reports sometimes mixed established measures, proposals,
recommendations, scenarios, and aspirations within the same narrative.

### Causal-language problems

| Causal-language problem | Reports | Percentage |
|---|---:|---:|
| YES | 5 | 38.5% |
| NO | 8 | 61.5% |

Five reports used language suggesting that one factor caused,
facilitated, increased, mitigated, or enabled another outcome without
sufficient outcome evidence.

For example, an intended reduction of bureaucratic barriers may be
described in the source as a proposed measure. A community report
should not automatically state that the measure has already increased
PV deployment.

### Unsupported elaboration

| Unsupported elaboration | Reports | Percentage |
|---|---:|---:|
| YES | 9 | 69.2% |
| NO | 4 | 30.8% |

Unsupported elaboration appeared in nine reports. Typical examples
included:

- combining separate statements into an apparently established causal
  narrative;
- assigning responsibilities to actors based only on association;
- interpreting policy intentions as implementation effects;
- treating a broad relationship as a specific operational mechanism;
- expanding limited qualitative evidence into general market
  conclusions.

## Interpretation

Pilot 2 community reports are more domain-specific and more useful
than the Pilot 1 reports. The controlled schema allows reports to
distinguish policies, technologies, targets, market metrics,
infrastructure, support schemes, stakeholders, and constraints.

The reports also show improved caution. Many explicitly acknowledge
that targets are scenario-based, proposed, planned, or unverified.

However, community reports remain the least authoritative stage of
the pipeline because they are produced after several transformations:

1. PDF conversion;
2. chunking;
3. entity extraction;
4. entity typing;
5. relationship extraction;
6. community detection;
7. LLM report generation.

An error or ambiguity introduced at an earlier stage can therefore
propagate into the community report. The report-generating LLM can
then introduce an additional layer of interpretation.

## Main conclusion

Pilot 2 community reports are suitable for:

- exploratory analysis;
- identifying major themes;
- providing an overview of graph communities;
- supporting question answering;
- locating potentially relevant entities and relationships.

They are not independently suitable for:

- factual verification;
- establishing causality;
- confirming policy implementation;
- proving measured market effects;
- serving as authoritative evidence without source tracing.

The original PDF and its text units must remain the authoritative
evidence. Every important generated claim should be traceable through:

`Community report → Relationship/entity → Text unit → Source document`

## Recommended treatment in the controlled KG

Community reports should not be imported as verified facts into the
controlled knowledge graph.

They should instead be stored as generated analytical artifacts with
metadata such as:

- source community identifier;
- generation model;
- generation date;
- grounding status;
- verification status;
- inherited-error indicator;
- modality-preservation assessment;
- source entity and relationship references.

Verified entities, relationships, targets, metrics, and policy claims
should remain separate from LLM-generated narrative summaries.

## Recommended improvements

1. Perform entity resolution before community detection and report
   generation.

2. Remove or correct unsupported and incorrectly directed
   relationships before generating reports.

3. Require report prompts to preserve explicit modality labels.

4. Prohibit causal language unless the supporting relationship is
   explicitly causal and source-supported.

5. Require every important claim to cite graph records and their
   supporting text units.

6. Distinguish measured outcomes from targets, scenarios,
   recommendations, proposals, and legal requirements.

7. Treat community-report rank as GraphRAG importance metadata, not as
   factual confidence.

8. Use community reports for navigation and synthesis, while using
   source-backed graph records for formal analysis and explanation.

## Overall assessment

Pilot 2 substantially improves the thematic usefulness of GraphRAG's
community reports, especially for policy, technical, and market
analysis. Nevertheless, the audit shows that only 30.8% of the sampled
reports were primarily accurate syntheses, while 69.2% contained new
LLM claims and 69.2% contained unsupported elaboration.

The main bottleneck is therefore no longer simply generic entity
typing. It is now the propagation of graph imperfections into
community reports, combined with additional interpretation during
LLM-based report generation.
