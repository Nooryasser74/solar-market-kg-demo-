```python
from pathlib import Path
import pandas as pd

project_root = Path(r"C:\Users\HP-ZBOOK i7\graphrag_test")

pilot1_units = pd.read_csv(
    project_root / "pilot_01" / "inspection" / "09_text_units_all.csv"
)

pilot2_units = pd.read_csv(
    project_root / "pilot_02" / "inspection" / "09_text_units_all.csv"
)

pilot1_units = pilot1_units.sort_values(
    "human_readable_id"
).reset_index(drop=True)

pilot2_units = pilot2_units.sort_values(
    "human_readable_id"
).reset_index(drop=True)

print("Pilot 1 text units:", len(pilot1_units))
print("Pilot 2 text units:", len(pilot2_units))

print(
    "Chunk texts identical:",
    pilot1_units["text"].tolist()
    == pilot2_units["text"].tolist()
)

print(
    "Token counts identical:",
    pilot1_units["n_tokens"].tolist()
    == pilot2_units["n_tokens"].tolist()
)

comparison = pd.DataFrame({
    "chunk": pilot1_units["human_readable_id"],
    "pilot1_tokens": pilot1_units["n_tokens"],
    "pilot2_tokens": pilot2_units["n_tokens"],
    "pilot1_characters": pilot1_units["text"].str.len(),
    "pilot2_characters": pilot2_units["text"].str.len(),
    "text_identical": (
        pilot1_units["text"] == pilot2_units["text"]
    )
})

comparison
```

## Pilot 2 Text-Unit Inspection — Conclusion

Pilot 1 and Pilot 2 each produced 17 text units. An exact comparison
showed that the text and token count of every corresponding text unit
were identical.

The first 16 text units contained 1,200 tokens each, while the final
text unit contained 124 tokens. The character counts were also identical
between the pilots for every chunk.

This confirms that Pilot 1 and Pilot 2 supplied the same source content
with the same chunk boundaries to the graph-extraction LLM. Therefore,
differences in entities, relationships, communities, or community reports
cannot be attributed to changes in chunk size, overlap, tokenization, or
chunk boundaries.

The chunking limitations identified in Pilot 1 remain present because
the chunking configuration was deliberately retained. These include
possible loss of context across boundaries and the short final chunk
containing mainly end matter. However, chunking is an experimental
control rather than a changed variable in Pilot 2.

The principal changed variables are therefore the controlled entity
types and the revised extraction and summarization prompts.
