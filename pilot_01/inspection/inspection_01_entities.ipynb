## Stage 3 — Preliminary Entity-Extraction Inspection

### Objective

The purpose of this stage is to evaluate whether GraphRAG extracted appropriate entities from the Austrian Photovoltaic Strategy and whether those entities were assigned suitable types.

The entity audit distinguishes between:

1. **Entity-identification quality** — whether GraphRAG identified a meaningful concept from the source.
2. **Entity-typing quality** — whether GraphRAG assigned an appropriate category to that concept.
3. **Entity-resolution quality** — whether different names for the same concept were merged.
4. **Description quality** — whether the LLM-generated entity description is supported by the source.

This distinction is important because an entity may exist in the source but still have the wrong type, an unsupported description, or an unresolved duplicate.

---

### Complete entity output

GraphRAG extracted **312 entities** from the single Austrian Photovoltaic Strategy document.

The raw type distribution was:

| Raw GraphRAG type |   Count |    Share |
| ----------------- | ------: | -------: |
| `ORGANIZATION`    |     142 |    45.5% |
| `EVENT`           |     121 |    38.8% |
| `PERSON`          |      25 |     8.0% |
| `GEO`             |      24 |     7.7% |
| **Total**         | **312** | **100%** |

The graph-extraction configuration allowed only four entity types:

```yaml
entity_types:
  - organization
  - person
  - geo
  - event
```

These categories do not adequately represent the concepts found in the Austrian solar-market document. Missing domain-relevant categories include:

* `POLICY`
* `LAW`
* `TECHNOLOGY`
* `INFRASTRUCTURE`
* `TARGET`
* `SUPPORT_SCHEME`
* `CONSTRAINT`
* `MARKET_METRIC`
* `STAKEHOLDER_GROUP`
* `PROCESS`

The large number of entities classified as `ORGANIZATION` or `EVENT` must therefore not be interpreted as evidence that the document primarily describes organizations and events. The distribution is strongly influenced by the limited configuration.

---

### Sampling method

Manually reviewing all 312 entities in detail would be time-consuming. A systematic sample was therefore constructed.

The sample combines:

* The **20 entities with the highest graph degree**
* A reproducible random sample of **10 entities from each raw GraphRAG type**
* Removal of duplicated rows appearing in both groups

A fixed random seed was used:

```python
random_state=42
```

This makes the sample reproducible.

After removing overlaps, the audit sample contains **57 unique entities**:

| Raw type       | Sample count |
| -------------- | -----------: |
| `ORGANIZATION` |           22 |
| `EVENT`        |           13 |
| `GEO`          |           12 |
| `PERSON`       |           10 |
| **Total**      |       **57** |

The sample includes both highly connected entities and less prominent entities from every available raw type. It therefore avoids reviewing only the most obvious or important examples.

---

### Meaning of GraphRAG fields

The entity table includes:

| Field           | Meaning                                                  |
| --------------- | -------------------------------------------------------- |
| `title`         | Entity name produced by GraphRAG                         |
| `type`          | Raw entity type assigned by GraphRAG                     |
| `description`   | LLM-generated summary of the entity                      |
| `text_unit_ids` | Text units associated with the entity                    |
| `frequency`     | Number of extraction occurrences combined for the entity |
| `degree`        | Number of graph relationships connected to the entity    |

`frequency` and `degree` are structural measurements. They are not measures of factual accuracy or confidence.

For example, `ÖSTERREICH` has a degree of 82. This means that GraphRAG connected Austria to many other entities. It does not establish that all 82 relationships are correct.

### Duplicate and alias problems

The preliminary inspection also reveals several likely cases where the same real-world concept was represented by multiple nodes.

#### Austrian Photovoltaic Strategy

Possible aliases include:

```text
ÖSTERREICHISCHE PHOTOVOLTAIK-STRATEGIE
PHOTOVOLTAIK-STRATEGIE
PV-STRATEGIE
```

These appear to refer to the same strategy and should probably map to one canonical entity.

#### Austrian federal ministry

Possible aliases include:

```text
BMK
BUNDESMINISTERIUM FÜR KLIMASCHUTZ, UMWELT, ENERGIE,
MOBILITÄT, INNOVATION UND TECHNOLOGIE
```

The abbreviation and full name should be connected through an alias relationship or normalized to one canonical organization.

#### Photovoltaic installations

Possible variants include:

```text
PV-ANLAGE
PV-ANLAGEN
PHOTOVOLTAIKANLAGEN
```

These may represent singular, plural, or language variants of the same basic concept. Their precise treatment may depend on whether the final KG distinguishes between the technology class and individual installations.

#### Integrated Austrian Network Infrastructure Plan

Possible variants include:

```text
INTEGRIERTER ÖSTERREICHISCHER NETZINFRASTRUKTURPLAN (ÖNIP)
INTEGRIERTER ÖSTERREICHISCHER NETZINFRASTRUKTURPLAN (NIP)
```

These require verification because `ÖNIP` is the abbreviation used for the Austrian plan, while `NIP` may be an inconsistent variant.

#### Johannes Hackner

Possible variants include:

```text
HACKNER, JOHANNES
JOHANNES HACKNER
```

These appear to be different name orders for the same person.

These cases indicate a **record-linkage/entity-resolution problem**. Adding more entity types alone will not automatically merge abbreviations, singular/plural forms, shortened names, or reversed personal names.

---

### Overgeneralized entities

Some extracted nodes describe broad categories rather than clearly bounded entities.

Examples include:

```text
BUND
BEVÖLKERUNG
NATUR
MITGLIEDSTAATEN
VERORDNUNGEN
ENERGIEWENDE
PV
```

Some of these concepts may still be useful in a controlled KG, but they require explicit modelling decisions.

For example:

* `PV` may be an alias for the technology `Photovoltaics`.
* `ENERGIEWENDE` may be a strategic process rather than an event.
* `BEVÖLKERUNG` may be represented as a stakeholder category.
* `VERORDNUNGEN` may be too general unless linked to specific regulations.

The audit should not automatically reject every broad concept. It should determine whether the concept supports the project’s purpose of explaining relationships among Austrian solar-market policy, market, and technical factors.

---

### Description-quality concerns

GraphRAG’s entity descriptions were generated by an LLM and may combine information from several chunks.

Some descriptions use broad or evaluative language such as:

* “at the forefront”
* “plays a crucial role”
* “comprehensive national strategy”
* “widely deployed”
* “central hub”
* “leading role”

Such language may be reasonable as a summary but must not automatically be treated as an explicit source statement.

An entity can therefore have:

* A correct name
* A correct or incorrect type
* A partially unsupported description

Description quality must be assessed separately from entity existence.

---

## Initial diagnosis

The preliminary evidence suggests:

1. GraphRAG can identify conventional named entities such as countries, cities, organizations, and named individuals.
2. It also identifies many relevant solar-market concepts.
3. The four generic entity types are insufficient for the document domain.
4. Many technologies, policies, targets, infrastructure elements, market concepts, and stakeholder groups are forced into `ORGANIZATION`, `EVENT`, `PERSON`, or `GEO`.
5. Abbreviations, shortened names, plural variants, and reordered person names are not reliably resolved.
6. LLM-generated descriptions may extend beyond what the source states explicitly.

The largest apparent problem at this stage is therefore not a complete failure to recognize concepts. It is the combination of:

* **Entity-type/schema failure**
* **Entity-resolution failure**
* **Description overgeneralization**

---

## Current status and limitation

This section reports preliminary patterns identified through inspection of the complete output and the reproducible 57-entity sample.

The individual sample rows have not yet received final source-based classifications. Therefore, no final error percentages are reported at this point.

The next step is to trace the sampled entities to their supporting `text_unit_ids` and classify them as:

* `CORRECT_AND_RELEVANT`
* `CORRECT_BUT_IRRELEVANT`
* `WRONG_TYPE`
* `DUPLICATE_OR_ALIAS`
* `OVERGENERALIZED`
* `UNSUPPORTED_OR_HALLUCINATED`

The completed classification will support approximate error proportions and a final assessment of entity-extraction quality.


````python
def add_review(
    title,
    source_supported,
    classification,
    expected_type,
    duplicate_or_alias_of="",
    description_problem="",
    pipeline_stage="",
    notes="",
):
    mask = entity_review["title"] == title

    entity_review.loc[mask, "source_supported"] = source_supported
    entity_review.loc[mask, "primary_classification"] = classification
    entity_review.loc[mask, "expected_type"] = expected_type
    entity_review.loc[mask, "duplicate_or_alias_of"] = duplicate_or_alias_of
    entity_review.loc[mask, "description_problem"] = description_problem
    entity_review.loc[mask, "likely_pipeline_stage"] = pipeline_stage
    entity_review.loc[mask, "audit_notes"] = notes
````

## Stage 3 — Completed Entity-Extraction Audit

### Audit scope

A reproducible sample of **57 entities** was reviewed from the complete GraphRAG output of 312 entities.

The sample combined:

* The 20 most highly connected entities
* Ten randomly selected entities from each raw GraphRAG type
* Removal of duplicated rows appearing in both groups
* A fixed random seed of `42`

Each sampled entity was evaluated according to:

* Source support
* Relevance to the intended solar-market KG
* Entity-type correctness
* Duplicate or alias status
* Degree of generalization
* Description quality
* Likely pipeline stage responsible for any problem

The original Austrian Photovoltaic Strategy and the associated text units were treated as the authoritative evidence.

## Description-quality problems

The `description` field is generated by an LLM. It is not copied directly from the source.

Several descriptions use broad language such as:

* “at the forefront”
* “plays a crucial role”
* “leading role”
* “widely deployed”
* “central hub”
* “comprehensive initiative”

These phrases may combine information from multiple chunks or introduce interpretation.

For example:

```text
Source evidence:
The document identifies the ministry as publisher and discusses policies in its area.

GraphRAG description:
The ministry is described as the principal driver responsible for developing,
executing, and overseeing a broad range of national strategies.
```

The description may be broadly plausible, but not every part is necessarily stated explicitly in the supporting text.

Description quality must therefore be audited independently from entity-name and entity-type correctness.

## Pipeline diagnosis

### Entity-identification failure

Entity identification was not the dominant problem in the reviewed sample. Most sampled concepts were at least partially connected to source language.

### Entity-type/schema failure

This was the dominant problem.

The configuration forced domain concepts into four inappropriate generic categories. It accounts directly for 28 primary `WRONG_TYPE` classifications and also contributes to several duplicate and overgeneralized entities.

### Entity-resolution failure

Abbreviations, shortened names, singular/plural variants, and reordered personal names were not consistently merged.

This created multiple nodes for the same underlying concept.

### Relevance-filtering failure

Front matter, references, publication information, and broad contextual concepts contributed some entities that are correct but of limited value to the intended KG.

### Description-summarization failure

## Methodological limitation

The reported percentages describe the **57-entity audit sample**, not a simple random estimate of all 312 entities.

The sample intentionally includes:

* Highly connected entities
* Stratified random entities from every raw type

Consequently, the percentages are appropriate for diagnosing pipeline weaknesses but should not be presented as exact population-wide error rates without reviewing a larger purely random sample.

The primary classifications are also mutually exclusive for counting purposes. Some entities have multiple simultaneous problems.

For example:

```text
PV-ANLAGEN
```

is both:

* Wrongly typed
* A duplicate/alias

It was assigned one primary classification while the additional problem was recorded in the audit notes and pipeline-stage fields.

## Stage 3 conclusion

GraphRAG generally identified concepts related to the Austrian Photovoltaic Strategy, with 48 of 57 sampled entities directly supported and nine partially supported.

However, successful identification did not translate into a controlled or reliable domain graph. Only 13 sampled entities were classified as both correct and relevant without a more important primary problem.

The dominant failure is entity typing. The generic categories `ORGANIZATION`, `PERSON`, `GEO`, and `EVENT` are inadequate for representing solar-market policies, technologies, infrastructure, targets, metrics, stakeholder groups, regulatory mechanisms, and environmental factors.

Entity resolution is a second major weakness. Important abbreviations and variants such as BMK, PV, strategy-name variants, and person-name variants remain separated.

The evidence supports treating GraphRAG as a useful but noisy entity-extraction layer. Experiment 2 should test domain-specific entity types and domain-relevant prompt examples. Nevertheless, a downstream controlled normalization and entity-resolution step will still be necessary even if raw extraction improves.
