````python
import re

# Distinctive German phrases expected for each source fact
fact_search_terms = {
    "P01": ["100 Prozent", "2030", "erneuerbaren Energiequellen"],
    "P02": ["11 TWh", "2030"],
    "P03": ["42,5 %", "45 %", "2030"],
    "P04": ["Umsatzsteuer", "35 kWp"],
    "P05": ["Elektrizitätswirtschaftsgesetz", "Netzkapazitäten"],
    "P06": ["Erneuerbaren-Ausbau-Beschleunigungsgesetz", "Verfahrensbeschleunigung"],
    "P07": ["Bau- und Raumordnung", "Naturschutz", "Bundesländer"],
    "P08": ["Bund-Länder-Dialog", "Förderstrategien"],
    "P09": ["Wohnungseigentumsgesetz", "Reihenhäusern"],
    "P10": ["lebendiges Dokument"],

    "M01": ["6,3 TWh", "2023"],
    "M02": ["2,5 GW", "2023"],
    "M03": ["Gigawatt Neuinstallation", "2022"],
    "M04": ["41 TWh", "2040"],
    "M05": ["15 TWh", "Gebäuden"],
    "M06": ["2 GW pro Jahr"],
    "M07": ["21 TWh", "2030"],
    "M08": ["19 TWh", "2020"],
    "M09": ["600 Mio. Euro", "2023"],
    "M10": ["knapp 10 %", "etwa 2 %"],

    "T01": ["Digitalisierung des Verteilnetzes"],
    "T02": ["Erzeugungsspitzen aus Photovoltaik", "nicht mehr vollständig"],
    "T03": ["Smartes lokales Energiemanagement", "Speicherung"],
    "T04": ["E-Mobilität", "Wärmepumpe", "Wasserstoff"],
    "T05": ["Ost-West Orientierungen", "vertikale PV-Anlagen"],
    "T06": ["Agri-PV-Anlagen", "Landnutzungseffizienz"],
    "T07": ["Freiflächenanlagen", "Biodiversitätsanlagen"],
    "T08": ["Gebäuden des Bundes", "technisch und wirtschaftlich"],
    "T09": ["Energiegemeinschaften", "Betriebsmonitoring"],
    "T10": ["Haltbarkeit", "Reparaturmöglichkeit", "Wiederverwertung"],
}


def normalize_text(value):
    value = str(value)

    # Repair words divided by PDF line-break hyphenation
    value = re.sub(r"(\w)-\s+(\w)", r"\1\2", value)

    # Normalize spaces and capitalization
    value = re.sub(r"\s+", " ", value)

    return value.casefold().strip()


# Prepare normalized text for each chunk
text_units_check = text_units.copy()

text_units_check["normalized_text"] = (
    text_units_check["text"].fillna("").map(normalize_text)
)


def inspect_text_unit_capture(fact_id):
    terms = fact_search_terms[fact_id]
    normalized_terms = [normalize_text(term) for term in terms]

    full_matches = []
    partial_matches = []

    for row_number, row in text_units_check.iterrows():
        chunk_text = row["normalized_text"]

        matched_terms = [
            original_term
            for original_term, normalized_term
            in zip(terms, normalized_terms)
            if normalized_term in chunk_text
        ]

        if len(matched_terms) == len(terms):
            full_matches.append(row_number + 1)

        elif matched_terms:
            partial_matches.append(
                {
                    "chunk": row_number + 1,
                    "matched_terms": matched_terms,
                }
            )

    if full_matches:
        capture = "YES"
        evidence_chunks = ", ".join(
            f"Chunk {number}" for number in full_matches
        )

    elif partial_matches:
        capture = "PARTIAL"
        evidence_chunks = "; ".join(
            f"Chunk {item['chunk']}: {', '.join(item['matched_terms'])}"
            for item in partial_matches
        )

    else:
        capture = "NO"
        evidence_chunks = ""

    return pd.Series(
        {
            "text_unit_capture": capture,
            "text_unit_evidence": evidence_chunks,
        }
    )


text_unit_results = recall_checklist["fact_id"].apply(
    inspect_text_unit_capture
)

recall_checklist["text_unit_capture"] = (
    text_unit_results["text_unit_capture"]
)

recall_checklist["text_unit_evidence"] = (
    text_unit_results["text_unit_evidence"]
)

print("Text-unit capture results:")
print(recall_checklist["text_unit_capture"].value_counts())

recall_checklist[
    [
        "fact_id",
        "category",
        "text_unit_capture",
        "text_unit_evidence",
        "source_fact",
    ]
]
````

````python
def classify_overall_capture(row):
    if row["text_unit_capture"] != "YES":
        return "MISSING_BEFORE_EXTRACTION"

    if (
        row["entity_capture"] == "FULL"
        and row["relationship_capture"] == "FULL"
    ):
        return "FULLY_STRUCTURED"

    if (
        row["entity_capture"] != "FULL"
        and row["relationship_capture"] == "FULL"
    ):
        return "RELATIONSHIP_TEXT_BUT_ENTITY_INCOMPLETE"

    if row["relationship_capture"] == "PARTIAL":
        return "PARTIALLY_STRUCTURED"

    if (
        row["relationship_capture"] == "MISSING"
        and row["entity_capture"] in ["FULL", "PARTIAL"]
    ):
        return "ENTITY_ONLY"

    return "MISSING_FROM_GRAPH"


recall_checklist["overall_classification"] = (
    recall_checklist.apply(
        classify_overall_capture,
        axis=1,
    )
)

print("Overall fact-level classifications:")
print(
    recall_checklist[
        "overall_classification"
    ].value_counts()
)

print()
print("Overall classifications by category:")
print(
    pd.crosstab(
        recall_checklist["category"],
        recall_checklist["overall_classification"],
    )
)
````

# Stage 8 — Recall and Missing-Information Audit

## Objective

The previous entity and relationship audits evaluated the correctness of information generated by GraphRAG. This stage examined the opposite direction:

> How much important information from the original Austrian Photovoltaic Strategy was successfully represented in the graph?

This is a checklist-based recall or coverage assessment. It does not measure the recall of every sentence in the 44-page document.

## Method

A reference checklist of 30 important source facts was manually constructed from the original PDF.

The checklist contained:

| Category                     | Number of facts |
| ---------------------------- | --------------: |
| Policy and regulatory        |              10 |
| Market and numerical         |              10 |
| Technical and implementation |              10 |
| **Total**                    |          **30** |

The checklist included major targets, policies, measurements, technical requirements and implementation measures.

Every fact was traced through the following pipeline:

```text
Original PDF
    ↓
GraphRAG text units
    ↓
```

## Pipeline diagnosis

The recall audit indicates the following information flow:

```text
Original source facts             30
        ↓
Facts preserved in text units     30
        ↓
Facts with complete entities       4
        ↓
Facts with complete relationships  2
        ↓
Fully structured facts             1
```

The major loss of structure occurred after chunking.

The dominant causes were:

1. Absence of entity types for targets and market metrics.
2. Failure to represent values, units and dates structurally.
3. Free-text relationships rather than controlled predicates.
4. Loss of qualifications such as “planned,” “where feasible,” “approximately,” or “by 2040.”
5. Fragmentation of one source fact across several generic entities and relationships.
6. Technical details being left inside source text rather than becoming graph concepts.

## Interpretation

GraphRAG demonstrated high textual coverage but low structured fact coverage.

The system usually received and retained important source information in its text units. It also frequently recognized the general topics involved. However, recognizing the topic was not sufficient to construct an explainable domain knowledge graph.

For the intended solar-market KG, the difference between these statements is important:

```text
Photovoltaics is connected to Austria.
```

and:

```text
Austria
    → HAS_TARGET
    → 41 TWh of PV generation by 2040
```

The first is a broad thematic association. The second is a structured, source-grounded and queryable policy fact.

Pilot 01 produced many relationships of the first kind and relatively few complete facts of the second kind.

## Methodological limitations

The 30 facts were deliberately selected as important policy, market and technical facts. They were not a simple random sample of every statement in the document.

Therefore, the results should be described as:

```text
checklist-based recall
```

or:

```text
sample-based structured coverage
```

They should not be presented as the exact recall of the complete 44-page source.

The assessment also distinguishes structured capture from textual mention. A value appearing inside a text unit, entity description or community report does not necessarily mean that it was represented as a queryable knowledge-graph fact.

## Stage 8 conclusion

The original PDF and text-unit stages preserved all 30 selected facts. Nevertheless, only one fact was fully represented through complete entities and a complete relationship.

The most severe false-negative problem concerns numerical targets and market indicators. None of the ten selected quantitative facts became a complete structured relationship.

The results indicate that PDF conversion and chunking were not the principal causes of low structured recall. The main losses occurred during entity identification, entity typing and relationship construction.

Pilot 02 should therefore test:

* explicit `TARGET` and `MARKET_METRIC` entity types;
* structured fields for value, unit, base year and target year;
* domain-specific examples in the extraction prompt;
* controlled relationship predicates;
* preservation of modal language and implementation conditions;
* provenance from every controlled fact to its supporting text unit.

The results support treating GraphRAG as a useful textual and thematic extraction layer, but not as a sufficiently controlled fact-extraction system without additional schema guidance and downstream validation.
