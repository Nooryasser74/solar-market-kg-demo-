````python
top_connected_relationships = relationships.nlargest(
    10, "combined_degree"
)

top_weight_relationships = relationships.nlargest(
    10, "weight"
)

random_relationships = relationships.sample(
    n=25,
    random_state=42
)

relationship_audit_sample = (
    pd.concat(
        [
            top_connected_relationships,
            top_weight_relationships,
            random_relationships,
        ]
    )
    .drop_duplicates(subset="id")
    .reset_index(drop=True)
)

print("Relationship audit sample size:", len(relationship_audit_sample))

relationship_audit_sample[
    [
        "source",
        "target",
        "description",
        "weight",
        "combined_degree",
        "text_unit_ids",
    ]
]
````

````python
relationship_review = relationship_audit_sample[
    [
        "id",
        "source",
        "target",
        "description",
        "weight",
        "combined_degree",
        "text_unit_ids",
    ]
].copy()

relationship_review["source_supported"] = ""
relationship_review["primary_classification"] = ""
relationship_review["direction_correct"] = ""
relationship_review["endpoints_correct"] = ""
relationship_review["description_problem"] = ""
relationship_review["likely_pipeline_stage"] = ""
relationship_review["audit_notes"] = ""

relationship_review.head()
````

````python
def add_relationship_review(
    row_number,
    source_supported,
    classification,
    direction_correct,
    endpoints_correct,
    description_problem="",
    pipeline_stage="",
    notes="",
):
    mask = relationship_review.index == row_number

    relationship_review.loc[mask, "source_supported"] = source_supported
    relationship_review.loc[mask, "primary_classification"] = classification
    relationship_review.loc[mask, "direction_correct"] = direction_correct
    relationship_review.loc[mask, "endpoints_correct"] = endpoints_correct
    relationship_review.loc[mask, "description_problem"] = description_problem
    relationship_review.loc[mask, "likely_pipeline_stage"] = pipeline_stage
    relationship_review.loc[mask, "audit_notes"] = notes
````

## Stage 4 — Relationship-Extraction Quality Audit

### Objective

The purpose of this stage was to determine whether relationships extracted by GraphRAG are supported by the Austrian Photovoltaic Strategy.

The original PDF and the associated text units were treated as authoritative evidence. GraphRAG’s relationship descriptions and weights were treated as model-generated outputs requiring independent verification.

The audit examined:

* Whether the source supports the relationship
* Whether the relationship is explicit or inferred
* Whether the description generalizes beyond the source
* Whether the source and target entities are appropriate
* Whether the direction is meaningful
* Whether references or co-occurrence were misinterpreted as substantive relationships

---

## Complete relationship output

GraphRAG extracted **414 relationships** from one source document.

Each relationship contains:

| Field             | Meaning                                         |
| ----------------- | ----------------------------------------------- |
| `source`          | First entity selected by GraphRAG               |
| `target`          | Second entity selected by GraphRAG              |
| `description`     | LLM-generated explanation of the connection     |
| `weight`          | Aggregated GraphRAG relationship-strength value |
| `combined_degree` | Combined graph connectivity of the endpoints    |
| `text_unit_ids`   | Text units associated with the relationship     |

Neither `weight` nor `combined_degree` represents factual confidence.

A high value does not demonstrate that a relationship is true. It may reflect repeated extraction, merged descriptions, frequent entities, overlap, or graph connectivity.

---

## Sampling method

A reproducible sample was created from three components:

* Ten relationships with the highest `combined_degree`
* Ten relationships with the highest GraphRAG `weight`
* Twenty-five randomly sampled relationships
* Removal of duplicated rows appearing in more than one component

A fixed random seed was used:

```python
random_state=42
```

After removing overlaps, the final audit sample contained **42 unique relationships**.

The sample includes:

* Structurally influential relationships
* Relationships given high weight by GraphRAG
* Ordinary randomly selected relationships

This design supports bottleneck diagnosis but is not a simple random sample of all 414 relationships.


```text
ÖSTERREICH → BUNDESLÄNDER
```

```text

```

The source explicitly identifies the ministry as publisher.

GraphRAG additionally describes the ministry as:

* Owner
* Principal driver
* Formulator
* Active implementation supervisor
* Coordinator of regulatory and funding mechanisms

Some of these roles may be plausible, but the complete description exceeds the explicit publication evidence.

### FFE and Austria

```text
FORSCHUNGSSTELLE FÜR ENERGIEWIRTSCHAFT → ÖSTERREICH
```

FFE appears as an author or contributor to a cited study.

### PV systems and the solar industry

```text
PV-ANLAGE → SOLARINDUSTRIE
```

The source connects the PV industry with domestic manufacturing and value creation.

The generated description combines manufacturing, implementation, innovation, domestic markets, and international markets into one broad relationship.

These cases demonstrate that GraphRAG frequently transforms narrower source statements into more comprehensive narratives.

### PV Austria and the Austrian PV Strategy

```text
BUNDESVERBAND PV AUSTRIA
    →
ÖSTERREICHISCHE PHOTOVOLTAIK-STRATEGIE
```

GraphRAG states that PV Austria helped shape and report on the Austrian PV Strategy.

The source cites PV Austria publications but does not explicitly state that the association shaped the strategy.

These examples show that references and co-occurrence can produce unsupported graph edges.

## Wrong endpoint

One sampled relationship contained a material endpoint error:

```text
ELWOG → ERNEUERBAREN-RICHTLINIE 2018/2001
```

The source explains that the new `ElWG` replaces the previous `ElWOG` and serves, among other things, to implement EU directives.

GraphRAG connected the old `ElWOG` to the implementation relationship instead of the new `ElWG`.

This is not merely a broad inference. It changes which law performs the stated function.

The likely cause is entity-name similarity combined with multiple related laws appearing in the same passage.

## Direction and relationship-schema limitations

GraphRAG’s raw relationships do not use controlled predicates such as:

```text
IMPLEMENTS
PUBLISHES
HAS_TARGET
SUPPORTS
CONSTRAINS
APPLIES_IN
FUNDED_BY
DERIVED_FROM
```

Instead, they contain a generic source, target, and free-text description.

This creates several problems:

1. The meaning of direction is sometimes unclear.
2. Reverse duplicates can coexist.
3. The same endpoints can carry different narrative descriptions.
4. Automated validation becomes difficult.
5. Relationships cannot be reliably queried by semantic type.

For example, both of the following appear:

```text
AUSTRIA → AUSTRIAN PV STRATEGY
```

```text
AUSTRIAN PV STRATEGY → AUSTRIA
```

Without a controlled predicate, it is unclear whether these mean:

* Austria has the strategy
* The strategy applies in Austria
* Austria created the strategy
* Austria benefits from the strategy
* The strategy concerns Austria

A controlled KG should replace these vague links with explicit predicates.

---

## Weight interpretation

Some relationships have high GraphRAG weights:

```text
AUSTRIA → AUSTRIAN PV STRATEGY
weight = 67
```

```text
AUSTRIAN PV STRATEGY → AUSTRIA
weight = 34
```

These values must not be interpreted as:

* Confidence percentages
* Truth probabilities
* Human validation scores
* Factual reliability measures

In the reviewed sample, high-weight relationships could still contain overgeneralized descriptions.

GraphRAG weight should therefore be treated as an internal structural or extraction-strength value, not as evidence.

---

## Main relationship-error mechanisms

### Causal generalization

The source says that policies support PV expansion. GraphRAG states that adoption is a direct result of those policies.

### Co-occurrence inference

Two actors appear in the same section. GraphRAG creates a specific operational relationship between them.

### Reference misinterpretation

Two organizations or authors appear together in a citation. GraphRAG creates a general collaboration, funding, or advisory relationship.

### Description expansion

A narrow relationship is converted into a long narrative containing roles, goals, effects, and causal claims.

### Entity confusion

Similar policy or law names cause GraphRAG to connect the relationship to the wrong entity, as in the `ElWG` versus `ElWOG` example.

### General-knowledge addition

The description includes correct background knowledge that is not explicitly provided by the supporting source passage.

## Severity assessment

**Direct relationship extraction:** Moderate quality
GraphRAG successfully extracts some explicit relationships.

**Inference control:** Poor to moderate quality
Many plausible relationships are not explicitly stated.

**Description precision:** Poor quality
Descriptions frequently expand beyond the narrow source evidence.

**Endpoint resolution:** Moderate concern
The `ElWG`/`ElWOG` example shows that similar entities can cause materially incorrect links.

**Reference handling:** Poor quality
Bibliographic co-occurrence can become unsupported domain relationships.

**Relationship typing:** Major limitation
Free-text relationships lack controlled semantic predicates.

## Methodological limitations

The 42-relationship sample is not a simple random sample of all 414 relationships.

It intentionally includes:

* Highly connected relationships
* High-weight relationships
* Random relationships

The reported proportions should therefore be used for pipeline diagnosis rather than treated as exact population-wide error rates.

The classification evaluates one primary problem per relationship. Some relationships have multiple simultaneous problems.

For example, a relationship may be:

* Partially supported
* Overgeneralized
* Directionally ambiguous
* Connected to an alias rather than a canonical entity

Only one primary classification was used for counting, while additional issues were recorded separately.

## Stage 4 conclusion

GraphRAG extracted several useful relationships from the Austrian Photovoltaic Strategy, particularly when the source stated the connection explicitly.

However, only 17 of 42 sampled relationships were classified as directly supported. The majority required qualification because GraphRAG introduced inference, broad causal language, generalized descriptions, reference-based associations, or an endpoint error.

The relationship descriptions are substantially less reliable than simple evidence-linked triples. GraphRAG often converts a narrow source connection into a broad narrative containing additional roles, goals, and effects.

For the controlled KG, raw GraphRAG relationships should not be accepted automatically. They should be transformed into:

```text
Canonical source entity
    →
Controlled predicate
    →
Canonical target entity
    →
Source document, page, and chunk
```

Explicit source relationships should be separated from inferred relationships. Inferred edges should carry a distinct status and must not be treated as equivalent to source facts.

The evidence indicates that relationship extraction is a significant bottleneck, particularly in relation to description generalization, reference interpretation, and lack of controlled predicates.
